```python
## Question 2
n=int(input("Enter a number: "))
m=1
## Method 1
while n>0:
    m=m*n
    n=n-1
print(m)
## Method 2 : Recursion
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(5))
```
